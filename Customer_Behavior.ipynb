## Dealing Null Values

```python
df.isnull().sum()
```

### size

```python
df['Size'].unique()
```

```python
df.groupby('Category')['Size'].unique()
```

```python
df.groupby('Item Purchased')['Size'].unique()
```

```python
# Handling null values and repalacing all sizes in Eletcronics Category as no size for electronics
df.loc[df['Category']=='Electronics','Size']="N/A"
```

```python
df.groupby('Category')['Size'].unique()
```

```python
# Handling null values and repalacing all sizes in Accessories Category no size for them
df.loc[df['Category']=='Accessories','Size']="N/A"
```

```python
df.groupby('Category')['Size'].unique()
```

```python
# Handling null values and repalacing all sizes in Footwear Category as no such size for Footwear
df.loc[df['Category']=='Footwear','Size']="Free Size"
```

```python
df.groupby('Category')['Size'].unique()
```

```python
# Handling null values in Clothing Category
df.loc[(df['Category']=='Clothing') & (df['Size'].isnull()) ,'Size']=df[df["Category"]=='Clothing']['Size'].mode()[0]
```

```python
df.groupby('Category')['Size'].unique()
```

```python
df.isnull().sum()
```

### Review Rating

```python
# mean of review rating
df['Review Rating'].mean()
```

```python
# mode of review rating
df['Review Rating'].mode()
```

```python
# median of review rating
df['Review Rating'].median()
```

```python
# mean of review rating item wise
df.groupby('Item Purchased')['Review Rating'].mean()
```

```python
# Fill null values of review with mean rating of each item
df['Review Rating']=df['Review Rating'].fillna(df.groupby('Item Purchased')['Review Rating'].transform('mean'))
```

```python
df.isnull().sum()
```

```python
 # decimal places correct
df['Review Rating']=df['Review Rating'].round(2)
```

```python
df.head()
```

```python
df.isnull().sum()
```

### Previous Purchase

```python
# put zero if not purchased anything
df['Previous Purchases']=df['Previous Purchases'].fillna(0)
```

```python
df.head()
```

```python
df.isnull().sum()
```

### Purchase Amount 

```python
# Fill null values of Purchase Amount (USD) with mean Purchase Amount (USD) of each item
df['Purchase Amount (USD)']=df['Purchase Amount (USD)'].fillna(df.groupby('Item Purchased')['Purchase Amount (USD)'].transform('mean'))
```

```python
df.isnull().sum()
```
